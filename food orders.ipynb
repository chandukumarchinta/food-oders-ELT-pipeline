### Task 3 — Design Your Cleaning Rules Before Changing Data


## 1. Data-Cleaning Rules

### 1. Exact Duplicate Rows
- **Handling Rule:** Remove exact duplicate rows, keeping the first occurrence.
- **Reason:** Prevent duplicate orders and incorrect revenue calculations.

### 2. Repeated Order ID
- **Handling Rule:** Investigate repeated IDs. Remove only confirmed duplicate order records and retain legitimate records.
- **Reason:** An Order ID should normally identify one order, but repeated IDs may need investigation.

### 3. Missing Order Value
- **Handling Rule:** Keep missing values as NaN and exclude these records from order-value calculations.
- **Reason:** Avoid assuming an unknown order value is zero.

### 4. Missing Delivery Fee
- **Handling Rule:** Keep missing values as NaN and exclude them from delivery-fee calculations.
- **Reason:** The actual delivery fee is unknown.

### 5. Missing Commission Fee
- **Handling Rule:** Keep missing values as NaN and exclude them from commission calculations.
- **Reason:** Avoid underestimating or miscalculating platform revenue.

### 6. Missing Payment Processing Fee
- **Handling Rule:** Keep missing values as NaN and exclude them from processing-fee calculations.
- **Reason:** The actual processing cost is unknown.

### 7. Missing Refunds/Chargebacks
- **Handling Rule:** Keep missing values as NaN and investigate the source.
- **Reason:** Missing refund information does not necessarily mean zero refunds.

### 8. Missing Payment Method
- **Handling Rule:** Replace missing values with `Unknown`.
- **Reason:** The payment method is unavailable.

### 9. Missing Discounts and Offers
- **Handling Rule:** Replace missing values with `Not Specified`.
- **Reason:** Missing offer details do not prove that no offer was applied.

### 10. Invalid Timestamps
- **Handling Rule:** Convert invalid timestamp values to NaT and investigate them.
- **Reason:** Invalid dates cannot be used reliably for time-based analysis.

### 11. Impossible Delivery Durations
- **Handling Rule:** Flag negative delivery durations and exclude invalid records from delivery-time analysis.
- **Reason:** Delivery cannot occur before an order is placed.

### Business Assumptions

1. Every Order ID should represent a unique order.
2. Missing financial values should not automatically be replaced with zero.
3. Unknown payment methods will be labelled as `Unknown`.
4. Missing discount information will be labelled as `Not Specified`.
5. Invalid timestamps will not be used in delivery-duration calculations.
6. Original raw data must be preserved for auditing and comparison.

### Conclusion

Cleaning rules are designed before modifying the dataset to ensure data accuracy, consistency, and reliable business analysis.

### Task 4 — Clean the Dataset

```python
# Create a copy of the original dataset
work_df = raw_df.copy()

print("Original Dataset Shape:", raw_df.shape)
print("Working Dataset Shape:", work_df.shape)
```

```python
work_df["Delivery Duration (Minutes)"] = (
    work_df["Delivery Date and Time"]
    - work_df["Order Date and Time"]
).dt.total_seconds() / 60

# Identify impossible delivery durations
invalid_duration = (
    work_df["Delivery Duration (Minutes)"] < 0
)

print("Impossible Delivery Durations:", invalid_duration.sum())

display(work_df[invalid_duration])
```

```python
work_df["Delivery Duration Status"] = np.where(
    invalid_duration,
    "Invalid",
    "Valid"
)

print(work_df["Delivery Duration Status"].value_counts())
```

### Task 5 — Transform the Data and Create Derived Columns

```python
# Create a copy of cleaned dataset
transform_df = work_df.copy()

print("Dataset Shape:", transform_df.shape)
```

```python
transform_df["Delivery_Time_Minutes"] = (
    transform_df["Delivery Date and Time"]
    - transform_df["Order Date and Time"]
).dt.total_seconds() / 60

display(transform_df[[
    "Order Date and Time",
    "Delivery Date and Time",
    "Delivery_Time_Minutes"
]].head())
```

```python
transform_df["Order_Date"] = (
    transform_df["Order Date and Time"].dt.date
)

display(transform_df[[
    "Order Date and Time",
    "Order_Date"
]].head())
```

```python
transform_df["Order_Hour"] = (
    transform_df["Order Date and Time"].dt.hour
)

display(transform_df[[
    "Order Date and Time",
    "Order_Hour"
]].head())
```

```python
revenue_columns = [
    "Commission Fee",
    "Delivery Fee",
    "Payment Processing Fee",
    "Refunds/Chargebacks"
]

# Check whether all four values are available
complete_revenue = transform_df[revenue_columns].notna().all(axis=1)

# Calculate platform net revenue
transform_df["Platform_Net_Revenue"] = np.where(
    complete_revenue,
    (
        transform_df["Commission Fee"]
        + transform_df["Delivery Fee"]
        - transform_df["Payment Processing Fee"]
        - transform_df["Refunds/Chargebacks"]
    ),
    np.nan
)

display(transform_df[revenue_columns + ["Platform_Net_Revenue"]].head(10))
```

```python
missing_delivery_time = transform_df[
    transform_df["Delivery_Time_Minutes"].isna()
]

print("Records Without Delivery Duration:",
      len(missing_delivery_time))

display(missing_delivery_time)
```

```python
negative_delivery = transform_df[
    transform_df["Delivery_Time_Minutes"] < 0
]

print("Negative Delivery Durations:", len(negative_delivery))

display(negative_delivery)
```

```python
long_delivery = transform_df[
    transform_df["Delivery_Time_Minutes"] > 1440
]

print("Delivery Durations Above 24 Hours:", len(long_delivery))

display(long_delivery)
```

```python
transform_df["Delivery_Status_Check"] = np.select(
    [
        transform_df["Delivery_Time_Minutes"].isna(),
        transform_df["Delivery_Time_Minutes"] < 0,
        transform_df["Delivery_Time_Minutes"] > 1440
    ],
    [
        "Missing Duration",
        "Invalid Negative Duration",
        "Unusually Long Duration"
    ],
    default="Valid"
)

print(transform_df["Delivery_Status_Check"].value_counts())
```

```python
print(
    "Missing Platform Net Revenue:",
    transform_df["Platform_Net_Revenue"].isna().sum()
)

print(
    "Available Platform Net Revenue:",
    transform_df["Platform_Net_Revenue"].notna().sum()
)
```

### Observations

1. **Delivery_Time_Minutes:** Calculated delivery duration by finding the difference between delivery timestamp and order timestamp in minutes.

2. **Order_Date:** Extracted the date portion from the order timestamp for daily order analysis.

3. **Order_Hour:** Extracted the hour from the order timestamp to understand customer ordering patterns.

4. **Platform_Net_Revenue:** Calculated using Commission Fee + Delivery Fee - Payment Processing Fee - Refunds/Chargebacks. Revenue is calculated only when all four financial values are available.

### Questions and Answers

**1. Which records cannot have a delivery duration calculated, and why?**

Records with missing or invalid order timestamps or delivery timestamps cannot have delivery duration calculated because both timestamps are required.

**2. Are any calculated delivery durations negative or implausible?**

Negative delivery durations indicate delivery occurred before the order was placed. Durations exceeding 24 hours are flagged for investigation rather than automatically deleted.

**3. Why should missing values in the revenue formula not automatically be replaced with zero?**

Missing financial values do not necessarily mean zero. Replacing them with zero could produce incorrect platform revenue and misleading business reports.

### Conclusion

The cleaned dataset has been transformed by creating delivery duration, order date, order hour, and platform net revenue columns. These derived columns help analyze delivery performance, customer ordering patterns, and platform revenue.
